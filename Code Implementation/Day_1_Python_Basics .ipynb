## Section 2 - Python String (30 mins)

### 2.1 String Basics

````python
# Defining a string
string_1 = "Hello, world!"
string_2 = 'Hello, world!'
print(string_1)
print(string_2)
````

````python
# Multiline String
long_text  = """This is how multiline 
string is defined."""
print(long_text)
````

````python
# Accessing by index
print(string_1[1])
print(string_1[0:5])
print(string_1[7:])
````

````python
# String concatination 
string_1 = "Hello,"
string_2 = "World!"

full_string = string_1 + string_2
print(full_string)
````

````python
full_string = string_1 + " " + string_2
print(full_string)
````

````python
# Using fstring 
full_string = f"{string_1} {string_2}"
print(full_string)
````
